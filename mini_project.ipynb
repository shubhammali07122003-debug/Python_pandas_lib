```python
import pandas as pd
```

```python

data = {
    'City': ['Pune', 'Mumbai', 'Delhi', 'Pune', 'Mumbai'],
    'Product': ['A', 'B', 'A', 'C', 'B'],
    'Sales': [100, 200, 150, 300, 250],
    'Quantity': [2, 3, 1, 5, 4]
}

df = pd.DataFrame(data)
df
```

```python
total_sales = df['Sales'].sum()
total_sales
```

```python
city_wise_sales = df.groupby('City')['Sales'].sum()
city_wise_sales
```

```python
df['revenue'] = df['Sales'] * df['Quantity']
df
```

```python
top_product = df.groupby('Product')['revenue'].sum()
top_product
```

```python
top_product.idxmax()
```

```python
df.groupby('City')['revenue'].sum()idxmax()
```
